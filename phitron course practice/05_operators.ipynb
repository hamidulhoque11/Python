```python
#Arithmetic operators +,-,*,/,%
a=int(input("Enter any value:"))
b=int(input("enter any value:"))
print("a=",a)
print("b=",b)
add=a+b
sub=a-b
mul=a*b
div=round(a/b,2)
rem=a%b
print("a+b:",add)
print("a-b:",sub)
print("a*b:",mul)
print("a/b:",div)
print("a%b:",rem)
```
